```python
import pandas as pd
import numpy as np

# Create zone mapping
nodes_df = pd.read_csv('data/networks/klagenfurt/nodes.csv')

zone_mapping = pd.DataFrame({
    'node_id': nodes_df['node_id'],
    'zone_id': ['klagenfurt_center'] * len(nodes_df)
})
zone_mapping.to_csv('data/zones/klagenfurt/klagenfurt/node_zone_info.csv', index=False)

# Create simple demand data (sample trips)
np.random.seed(42)
demand_data = []
for i in range(100):  # 100 sample trips
    origin = np.random.choice(nodes_df['node_id'])
    destination = np.random.choice(nodes_df['node_id'])
    # Trip time between 7 AM (25200) and 9 AM (32400)
    trip_time = np.random.randint(25200, 32400)

    demand_data.append({
        'request_id': i,
        'origin_node': origin,
        'destination_node': destination,
        'earliest_pickup_time': trip_time,
        'latest_pickup_time': trip_time + 600  # 10 min window
    })

demand_df = pd.DataFrame(demand_data)
demand_df.to_csv('data/demand/klagenfurt/matched/klagenfurt/demand_requests.csv', index=False)
print(f"Created {len(demand_df)} sample trip requests")
```
