```python
import geopandas as gpd
from shapely.geometry import box

# Load the Austrian network
gdf = gpd.read_file("data/gip_network_ogd.gpkg")

# Define Klagenfurt bounding box (approximate)
# Klagenfurt coordinates: 46.624°N, 14.307°E
klagenfurt_bbox = box(14.25, 46.58, 14.36, 46.67)  # min_lon, min_lat, max_lon, max_lat

# Filter network data within Klagenfurt area
klagenfurt_network = gdf[gdf.geometry.intersects(klagenfurt_bbox)]

# Save filtered network
klagenfurt_network.to_file("klagenfurt_network.gpkg", driver="GPKG")
print(f"Extracted {len(klagenfurt_network)} network segments for Klagenfurt")
```
